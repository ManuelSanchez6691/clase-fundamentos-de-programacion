```python
celsius = float(input("Ingresa los grados Celsius: "))

fahrenheit = (celsius * 9/5) + 32

print("La temperatura en Fahrenheit es:", fahrenheit, "°F")
```

```python
kilometros = float(input("Ingresa los kilómetros: "))

millas = kilometros * 0.621371

print("La distancia en millas es:", millas)
```

```python
cena = float(input("Ingresa el costo de la cena: "))
porcentaje = float(input("Ingresa el porcentaje de propina: "))

propina = cena * (porcentaje / 100)
total = cena + propina

print("La propina es:", propina)
print("El total a pagar es:", total)
```
